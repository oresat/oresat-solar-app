```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import sys

def index_down(csv, val):
    least = csv.iloc[0].iloc[0] #first row, nm
    for idx in range(len(csv)):
        row = csv.iloc[idx]
        if (row.iloc[0] >= val):
            return least
        least = row

def extract_gw_irr(csv, outlist):
    for idx in range(len(outlist)):
        if (idx == 0):
            continue
        outlist[idx] = csv.iloc[idx].iloc[2]

def import_irr(csvpath):
    csvdata = pd.read_csv(csvpath, sep=' ')
    outlist = [0] * 2048
    extract_gw_irr(csvdata, outlist)
    return outlist

def calculate_error(arr1, arr2):
    if (len(arr1) != len(arr2)):
        raise("arguments not equal length")
    outlist = [0] * len(arr1)
    for idx in range(len(arr1)):
        # outlist[idx] = np.sqrt(np.abs(np.pow(arr1[idx], 2) - np.pow(arr2[idx], 2)))
        outlist[idx] = abs(arr1[idx] - arr2[idx])
    return outlist


sun = import_irr("data/sun.IRR")
sim_full = import_irr("data/SolarSimV2.0/fixed_preset_1.0.IRR")
sim_cyan = import_irr("data/SolarSimV2.0/cyan_100p.IRR")
sim_halogen = import_irr("data/SolarSimV2.0/halogen_100p.IRR")
sim_violet = import_irr("data/SolarSimV2.0/violet_100p.IRR")
sim_white = import_irr("data/SolarSimV2.0/white_100p.IRR")
sim_match = import_irr("data/SolarSimV2.0/paramsv1.IRR")

print(len(sun))
reflen = 2048
nanometers = [0] * reflen
irradiance_err = [0] * reflen
AM0 = [0] * reflen
AM15 = [0] * reflen

refdata = pd.read_csv("data/sun.IRR", sep=' ') #just to extract the nanometers scale

#wavelengths start to overlap at 265.5 nm, row 149 of sun data
AM0_irradiance = pd.read_csv("data/AM0.csv")
AM15_irradiance = pd.read_csv("data/AM1.5.csv")
for idx in range(len(refdata)):
    if (idx == 0):
        continue
    row = refdata.iloc[idx]
    nanometers[idx] = row.iloc[0]
    AM0[idx] = index_down(AM0_irradiance, row.iloc[0]).iloc[1]
    # print("Err: ", np.sqrt(pow(row.iloc[2], 2) + pow(index_down(sun_irradiance, row.iloc[0]).iloc[1], 2)))
    # AM0_err[idx] = np.sqrt(pow(row.iloc[2], 2) + pow(index_down(AM0_irradiance, row.iloc[0]).iloc[1], 2))


print("done")
```

```python

# print(sim_irr)
# print(sun_irr)
plt.title("solar irradiance, simulator vs AM0 standard solar spectra")
plt.plot(nanometers, sim_full, label="Simulator")
plt.plot(nanometers, AM0, label="AM0")
plt.plot(nanometers, sun, label="Sun measured")
plt.ylabel("$W*m^{-2}*nm^{-1}$")
plt.xlabel("wavelength in $nm$")
plt.legend()
plt.grid(True)
plt.xlim(260,)
plt.show()

plt.title("different light sources")
plt.plot(nanometers, sim_cyan, label="cyan led")
plt.plot(nanometers, sim_violet, label="violet led")
plt.plot(nanometers, sim_white, label="white led")
plt.plot(nanometers, sim_halogen, label="halogen bulb")
sim_sum = np.add(np.add(sim_cyan, sim_violet), np.add(sim_white, sim_halogen))
# plt.plot(nanometers, sim_sum, label="sum")
# plt.plot(nanometers, sim_full, label="everything on")
# plt.plot(nanometers, AM0, label="AM0")
plt.ylabel("$W*m^{-2}*nm^{-1}$")
plt.xlabel("wavelength in $nm$")
plt.legend()
plt.grid(True)
plt.xlim(260,)
plt.show()
```

```python
from scipy.optimize import curve_fit

def fit_func(nm, violet, cyan, white, hal):
    nm = nm.astype(int)
    # print(nm)
    sim_sum = np.add(
        np.add(
            np.multiply(np.array(sim_cyan)[nm], cyan),
            np.multiply(np.array(sim_violet)[nm], violet)
        ),
        np.add(
            np.multiply(np.array(sim_white)[nm], white), 
            np.multiply(np.array(sim_halogen)[nm], hal)
        )
    )
    return sim_sum
    # return sim_violet[nm] * violet + sim_cyan[nm] * cyan + sim_white[nm] * white + sim_halogen[nm] * halogen


# params = [1, 1, 1, 1]
# print(sim_cyan)
nmidx = np.arange(2048, dtype=np.int16)
# print(nmidx)
# print(fit_func(51, 1, 1, 1, 1))
popt, pcov = curve_fit(fit_func, nmidx, AM0)
# print(popt)
error = calculate_error(AM0, fit_func(nmidx, *popt))
print("violet: ", popt[0], " cyan: ", popt[1], " white: ", popt[2], " hal: ", popt[3])


plt.title("error over spectrum")
plt.plot(nanometers, AM0, label="AM0")
# plt.plot(nanometers, sim_sum, label="sum")
plt.plot(nanometers, fit_func(nmidx, *popt), label="best sim can do")
plt.plot(nanometers, error, label="error")
plt.ylabel("$W*m^{-2}*nm^{-1}$")
plt.xlabel("wavelength in $nm$")
plt.legend()
plt.grid(True)
plt.xlim(250,)
plt.show()
plt.show()

plt.title("calculated match vs observed match")
plt.plot(nanometers, fit_func(nmidx, *popt), label="calculated")
plt.plot(nanometers, sim_match, label="observed")
# plt.plot(nanometers, error, label="error")
plt.ylabel("$W*m^{-2}*nm^{-1}$")
plt.xlabel("wavelength in $nm$")
plt.legend()
plt.grid(True)
plt.xlim(250,)
plt.show()
plt.show()

old_fit = fit_func(nmidx, *popt)


```

```python
#import different halogen intensities

hal15p = import_irr("data/SolarSimV2.0/hal_15p.IRR")
hal25p = import_irr("data/SolarSimV2.0/hal_25p.IRR")
hal30p = import_irr("data/SolarSimV2.0/hal_30p.IRR")
hal40p = import_irr("data/SolarSimV2.0/hal_40p.IRR")
hal45p = import_irr("data/SolarSimV2.0/hal_45p.IRR")
hal50p = import_irr("data/SolarSimV2.0/hal_50p.IRR")
hal60p = import_irr("data/SolarSimV2.0/hal_60p.IRR")
hal75p = import_irr("data/SolarSimV2.0/hal_75p.IRR")
hal80p = import_irr("data/SolarSimV2.0/hal_80p.IRR")
hal90p = import_irr("data/SolarSimV2.0/hal_90p.IRR")

hal_vals = [
    hal15p,
    hal25p,
    hal30p,
    hal40p,
    hal45p,
    hal50p,
    hal60p,
    hal75p,
    hal80p,
    hal90p,
    sim_halogen,
]

hal_idx = {
    15: 0,
    25: 1,
    30: 2,
    40: 3,
    45: 4,
    50: 5,
    60: 6,
    75: 7,
    80: 8,
    90: 9,
    100: 10,
}

plt.title("different halogen intensities")
plt.ylabel("$W*m^{-2}*nm^{-1}$")
plt.xlabel("wavelength in $nm$")
plt.plot(nanometers, hal_vals[hal_idx[15]], label="15%")
plt.plot(nanometers, hal_vals[hal_idx[25]], label="25%")
plt.plot(nanometers, hal_vals[hal_idx[30]], label="30%")
plt.plot(nanometers, hal_vals[hal_idx[40]], label="40%")
plt.plot(nanometers, hal_vals[hal_idx[45]], label="45%")
plt.plot(nanometers, hal_vals[hal_idx[50]], label="50%")
plt.plot(nanometers, hal_vals[hal_idx[60]], label="60%")
plt.plot(nanometers, hal_vals[hal_idx[75]], label="75%")
plt.plot(nanometers, hal_vals[hal_idx[80]], label="80%")
plt.plot(nanometers, hal_vals[hal_idx[90]], label="90%")
plt.plot(nanometers, hal_vals[hal_idx[100]], label="100%")
plt.legend()


#gonna have, nanometers, intensity  +  11 arrays
# pick the arrays above/below given intensity
# lerp between points at that wavelength


#uses hal_vals and hal_idx
def arrlerp(intensity):
    #iterate through arrays
    #find if intensity is <min or >max or between two
    small = 0
    big = 0
    for compareto in list(hal_idx.keys()):
        if intensity > compareto:
            small = compareto
        elif intensity == compareto:
            small = compareto
            big = compareto
            break
        elif intensity < compareto:
            big = compareto
            break
    return (small, big)
    

#commanded intensity vs light intensity

#breaks below 15,
def pointlerp(arr_a, arr_b, wavelength, intensity):

    #1 get light intensities of arr_a and arr_b at wavelength
    b_intensity = hal_vals[hal_idx[arr_b]][wavelength]
    if arr_a == 0:
        a_intensity = 0
    else:
        a_intensity = hal_vals[hal_idx[arr_a]][wavelength]
  

        
    if arr_a == arr_b:
        return a_intensity
    
    #2 use (command?)intensity to lerp between them
    slope = (b_intensity - a_intensity) / (arr_b - arr_a)#in intensity/intensity command
    # print("slope is: ", slope)
    # print("intensity", a_intensity, b_intensity)
    #return lerped intensity
    #estimate = light intensity of lower bound + slope * intensity command difference
    int_est = a_intensity + slope * (intensity - arr_a)
    # print(a_intensity, "blah", b_intensity)
    # print(slope)
    # print(intensity)
    
    return int_est

# a, b = arrlerp(14)
# print(nanometers[1000])
# print(a, b)
# print(pointlerp(a, b, 1000, 14))




```

```python
from scipy.optimize import curve_fit

def fit_func(nm, violet, cyan, white, hal):
    nm = nm.astype(int)
    # print(nm)

    interpolated_hal = [0] * 2048
    a, b = arrlerp(hal)
    for idx, wavelength in enumerate(nm):
        interpolated_hal[idx] = pointlerp(a, b, wavelength, hal)
    
    sim_sum = np.add(
        np.add(
            np.multiply(np.array(sim_cyan)[nm], cyan),
            np.multiply(np.array(sim_violet)[nm], violet)
        ),
        np.add(
            np.multiply(np.array(sim_white)[nm], white), 
            # np.multiply(np.array(sim_halogen)[nm], hal)
            interpolated_hal
        )
    )
    return sim_sum
    # return sim_violet[nm] * violet + sim_cyan[nm] * cyan + sim_white[nm] * white + sim_halogen[nm] * halogen


# params = [1, 1, 1, 1]
# print(sim_cyan)
nmidx = np.arange(2048, dtype=np.int16)
# print(nmidx)
# print(fit_func(51, 1, 1, 1, 1))
popt, pcov = curve_fit(fit_func, nmidx, AM0)
# print(popt)
error = calculate_error(AM0, fit_func(nmidx, *popt))
print("violet: ", popt[0], " cyan: ", popt[1], " white: ", popt[2], " hal: ", popt[3])


plt.title("error over spectrum")
plt.plot(nanometers, AM0, label="AM0")
# plt.plot(nanometers, sim_sum, label="sum")
plt.plot(nanometers, fit_func(nmidx, *popt), label="best sim can do")
plt.plot(nanometers, error, label="error")
plt.ylabel("$W*m^{-2}*nm^{-1}$")
plt.xlabel("wavelength in $nm$")
plt.legend()
plt.grid(True)
plt.xlim(250,)
plt.show()
plt.show()

plt.title("calculated match vs observed match")
plt.plot(nanometers, fit_func(nmidx, *popt), label="calculated")
plt.plot(nanometers, sim_match, label="observed")
# plt.plot(nanometers, error, label="error")
plt.ylabel("$W*m^{-2}*nm^{-1}$")
plt.xlabel("wavelength in $nm$")
plt.legend()
plt.grid(True)
plt.xlim(250,)
plt.show()
plt.show()

plt.title("old vs new fit")
plt.plot(nanometers, fit_func(nmidx, *popt), label="new fit")
plt.plot(nanometers, old_fit, label="old fit")
plt.plot(nanometers, sim_match, label="observed")
# plt.plot(nanometers, error, label="error")
plt.ylabel("$W*m^{-2}*nm^{-1}$")
plt.xlabel("wavelength in $nm$")
plt.legend()
plt.grid(True)
plt.xlim(250,)
plt.show()
plt.show()
```
